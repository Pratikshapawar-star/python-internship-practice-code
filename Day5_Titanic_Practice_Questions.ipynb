## **A. NumPy (5 Questions)**
1. Create a NumPy array of the `Age` column and calculate its **mean**, **median**, and **standard deviation**.

3. Find the **index positions** where passenger ages are greater than 60.
4. Replace all missing values in the `Age` column with the **mean age** using NumPy.

```python
import numpy as np
```

```python
age_array = np.array(df['Age'].dropna())

age_mean = np.mean(age_array)
age_median = np.median(age_array)
age_std = np.std(age_array)

print("Mean Age:",age_mean)
print("Mean Age:",age_median)
print("Mean Age:",age_std)
```

```python
random_ages = np.random.randint(1,81,size=10)
print("Random Ages:",random_ages)
```

```python
indices = df[df['Age'] > 60].index

print("Index positions where Age > 60:")
print(list(indices))

```

```python
mean_age = np.nanmean(df['Age'])
df['Age'] = df['Age'].fillna(mean_age)

print(df['Age'].head())
```

```python
fare_age_array = df[['Fare', 'Age']].to_numpy()
print(fare_age_array[:5])
correlation_matrix = np.corrcoef(fare_age_array[:, 0], fare_age_array[:, 1])
correlation = correlation_matrix[0, 1]

print("Correlation between Fare and Age:", correlation)
```

Data cleaning and missing values

```python
missing_values = df.isnull().sum()
print(missing_values)
```

```python
df = df.drop(columns=['Cabin'])
print(df.head())
```

```python
median_age = df['Age'].median()
df['Age'] = df['Age'].fillna(median_age)
print(median_age)
```

Encoding and transformation

```python
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
print(df.head())
```

```python
df = pd.get_dummies(df, columns=['Embarked'], prefix='Embarked')
print(df.head())
```

```python
df['Fare_Normalized'] = (df['Fare'] - df['Fare'].min()) / (df['Fare'].max() - df['Fare'].min())
print(df.head())
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
```

Visualization with Matplotlib

```python
import matplotlib.pyplot as plt

plt.hist(df['Age'].dropna() ,bins=20)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
df['Pclass'].value_counts().sort_index().plot(kind='bar')
plt.title('Number of Passengers in Each Pclass')
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.show()
```

```python
plt.plot(df['PassengerId'], df['Fare'])
plt.title('Fare Trend by PassengerId')
plt.xlabel('PassengerId')
plt.ylabel('Fare')
plt.show()
```

```python
plt.scatter(df['Age'], df['Fare'])
plt.title('Fare vs Age')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

Visualization with Seaborn

```python
import seaborn as sns


sns.countplot(data=df, x='Sex')
plt.title('Passenger Count by Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.show()

```

```python

sns.boxplot(data=df, x='Pclass', y='Age')
plt.title('Age Distribution by Pclass')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.show()

```

```python
corr = df.corr(numeric_only = True)
plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()

```

```python

sns.violinplot(data=df, x='Survived', y='Fare')
plt.title('Fare Distribution by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Fare')
plt.show()
```
