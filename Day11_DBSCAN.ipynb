```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn import metrics
from sklearn.cluster import KMeans
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from sklearn.metrics import silhouette_samples
```

```python
data=pd.read_csv('/content/environmental_socioeconomic.csv')
```

```python
from sklearn.datasets import make_moons
x,y = make_moons(n_samples=300,noise=0.05,random_state=0)
x_scaled = StandardScaler().fit_transform(x)

plt.scatter(x_scaled[:,0],x_scaled[:,1],s=20)
plt.title('Original Data Distribution')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.grid(True)
plt.show()
```

```python
kmeans=KMeans(n_clusters=2,random_state=42)
kmeans.fit_predict(x_scaled)
```

```python
#Plot K-Means clustered data

clusters = kmeans.fit_predict(x_scaled)
plt.scatter(x_scaled[:,0],x_scaled[:,1],c=clusters,cmap = 'viridis',s=30)
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1],
            s=200,c='red',marker='*',label='Centriods')
plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True)
plt.show()
```

```python
sc= metrics.silhouette_score(x,clusters)
print('Silhouette Coefficient:%0.2f'% sc)
ari=metrics.adjusted_rand_score(y,clusters)
print('Adjusted Rand Index:%0.2f'% ari)
```

```python
db=DBSCAN(eps=0.3,min_samples=5)
clusters=db.fit_predict(x_scaled)

plt.scatter(x_scaled[:,0],x_scaled[:,1],c=clusters,cmap='viridis',s=30)
plt.title("DBSCAN Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True)
plt.show()
```

```python
from sklearn import metrics
sc=metrics.silhouette_score(x,clusters)
print('Silhouette Coefficient:%0.2f'% sc)
ari=metrics.adjusted_rand_score(y,clusters)
print('Adjusted Rand Index:%0.2f'% ari)
```
