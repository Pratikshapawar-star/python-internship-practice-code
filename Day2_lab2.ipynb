```python
# Import NumPy and Pandas
import pandas as pd
import numpy as np
```

```python
pd.__version__
```

```python
empty_series = pd.Series(dtype= "float64")
print(empty_series)
```

```python
arr=np.array([22,33,44,55,66])
number_series = pd.Series(arr)
print(number_series)
```

```python
number_series = pd.Series([22,33,44,55,66], index=["a","b","c","d","e"], name="Number")
print(number_series)
```

```python
number_list=[22,33,44,55,66]
number_series=pd.Series(number_list,index=["a","b","c","d","e"], name="Number")
print(number_series)
```

```python
number_dict={"a":22,"b":33,"c":44,"d":55,"e":66}
number_series=pd.Series(number_dict,name="number")
print(number_series)
```

```python
empty_df=pd.DataFrame(columns=["Column 1","Column 2","Column 3"])
print(empty_df)
```

```python
std_dict= {'Name': ['Tom','Nick','Kish','Jack'],'Age':[20,21,19,18]}
df=pd.DataFrame(std_dict)
print(df)
```

```python
list_of_lists=[["Tom",20],["Nick",21],["Krish",19],["Jack",18]]

df=pd.DataFrame(list_of_lists,columns=["Name","Age"])
print(df)
```

```python
list_of_dicts= [
    {"Name":"Amar","Age":29, "City":"Ahmedabad"},
    {"Name":"Akbar","Age":35, "City":"New Delhi"}

]
df=pd.DataFrame(list_of_dicts)
print(df)
```

```python
#creating array using random module
arr=np.random.randint(1,10, size=(3,5))

df=pd.DataFrame(arr,columns=["A","B","C","D","E"])
print(df)
```

```python
#create DataFrame
df=pd.DataFrame({
    "A": [1,2,3],
    "B":[4,5,6]
},index=['a','b','c'])
print(df)
```

```python
#reindex the DataFrame by using reindex method
new_index= ["c","a","e"]
df_reindexed = df.reindex(new_index)
print(df_reindexed)
```
