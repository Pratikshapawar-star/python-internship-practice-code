```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('/content/green transportation system.csv')
```

```python
data.head()
```

```python
data.shape
```

```python
data.isnull().sum()
```

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
data_scaled=scaler.fit_transform(data)

scaled_data=pd.DataFrame(data_scaled,columns=data.columns)
scaled_data.head()
```

```python
from sklearn.decomposition import PCA

pca=PCA()

pca.fit(data_scaled)
print('Explained varianc ratio for each principal component:')
print(pca.explained_variance_ratio_)

print('\nCumulative explained variance:')
print(np.cumsum(pca.explained_variance_ratio_))
```

```python
plt.figure(figsize=(8,6))
plt.bar(range(1,len(pca.explained_variance_ratio_)+1),pca.explained_variance_ratio_)
plt.title('Explained variance by each principal component')
plt.xlabel('Principal component')
plt.ylabel('Explained variance ratio')
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.plot(range(1, len(pca.explained_variance_ratio_)+1),marker='o',linestyle='--')
plt.title('Cumulative explained variance')
plt.xlabel('Number of principal components')
plt.ylabel('Cumulative explained variance')
plt.show()
```

```python
df=pd.DataFrame(dataset.data,columns=dataset.feature_names)
df.head()
```

```python
dataset.target
```

```python
df.describe()
```

```python
x=df
y=dataset.target
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
x_scaled
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x_scaled, y,test_size=0.2,random_state=30)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
model.score(x_test,y_test)
```

```python
from sklearn.decomposition import PCA

pca = PCA(0.95)
x_pca = pca.fit_transform(x)
x_pca.shape
```

```python
pca.explained_variance_ratio_
```

```python
pca.n_components_
```

```python
x_pca
```

```python
x_train_pca, x_test_pca, y_train, y_test = train_test_split(x_pca,y,test_size=0.2,random_state=30)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(x_train_pca,y_train)
model.score(x_test_pca,y_test)
```

```python
pca=PCA(n_components=4)
x_pca=pca.fit_transform(x)
x_pca.shape
```
